```python
"""
Author: Kelsey
Step 2: Ingestion Logic for SQL Server
Description: This script connects to the 'week4' database on the 
             Kelsey server to populate the schema.
"""

import pyodbc
import pandas as pd

# 1. Establish connection to your SQL Server
# Note: Trusted_Connection=yes uses your Windows login credentials.
conn_str = (
    r'DRIVER={ODBC Driver 17 for SQL Server};'
    r'SERVER=Kelsey;'
    r'DATABASE=Week 4;'
    r'Trusted_Connection=yes;'
)

try:
    conn = pyodbc.connect(conn_str)
    cursor = conn.cursor()
    print("✅ Successfully connected to Kelsey server.")

    # 2. Populate Truth Tables (Using T-SQL syntax)
    # SQL Server doesn't need "PRAGMA foreign_keys" as they are always on.
    
    regions = [('Alberta',), ('British Columbia',)]
    cursor.executemany("INSERT INTO Regions (RegionName) VALUES (?)", regions)

    categories = [('Camping',), ('Cycling',)]
    cursor.executemany("INSERT INTO Categories (CategoryName) VALUES (?)", categories)

    # 3. Insert specific records
    cursor.execute("INSERT INTO Customers (CustomerName, JoinDate) VALUES (?, ?)", 
                   ('John Doe', '2025-01-01'))

    conn.commit()
    print("✅ Database 'Week 4' Populated Successfully.")

except Exception as e:
    print(f"❌ Error: {e}")
    conn.rollback()
finally:
    cursor.close()
```
