## Step 2

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split

# 1. Create a complex, non-linear classification dataset
np.random.seed(42)
X = np.random.uniform(-1, 1, (1500, 2))

# Target Logic: 1 if inside a circle of radius 0.7, else 0
# This creates a "Failure Zone" that simple models struggle to map
y = (np.sqrt(X[:, 0]**2 + X[:, 1]**2) < 0.7).astype(int)

# 2. Add "Noise" (Simulating sensor glitches or Survivorship Bias)
# We flip 5% of the labels to make the model "work for it"
noise_idx = np.random.choice(len(y), 75, replace=False)
y[noise_idx] = 1 - y[noise_idx]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Visualize the challenge
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm', alpha=0.5)
plt.title("The Nonlinear Challenge: Can the model find the circle?")
plt.show()
```

## Step 3

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score, classification_report

# Initialize the competitors
models = {
    "Single Decision Tree (Weak)": DecisionTreeClassifier(max_depth=3),
    "Random Forest (Bagging)": RandomForestClassifier(n_estimators=100, max_depth=10),
    "XGBoost (Boosting)": XGBClassifier(n_estimators=100, learning_rate=0.1, max_depth=5)
}

# Run the shootout
results = {}
for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    results[name] = acc
    print(f"--- {name} ---")
    print(f"Accuracy: {acc:.4f}\n")

# Visualize the Results
plt.bar(results.keys(), results.values(), color=['red', 'blue', 'green'])
plt.ylabel("Accuracy Score")
plt.title("The Ensemble Jump")
plt.ylim(0.7, 1.0) # Zoom in to see the difference
plt.show()
```
